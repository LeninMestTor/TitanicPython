```python
import pandas as pd

df = pd.read_csv('titanic.csv')

df.head()
```

# 1. ¿Cuántos pasajeros hay en el dataset y cuántos PassengerId distintos aparecen? ¿Coinciden los dos resultados?

Se analiza en la tabla las columnas Passengerld y Name para devolver los elementos de cada una de estas

```python
total_filas = len(df)
total_ids = df['PassengerId'].nunique()

son_iguales = total_filas == total_ids

tabla_p1 = pd.DataFrame({
    "Indicador": ["Total Registros", "IDs Únicos", "Coinciden"],
    "Resultado": [total_filas, total_ids, son_iguales]
})

display(tabla_p1)
```

# 2. ¿Cuántos pasajeros de cada clase aparecen en el archivo? ¿Qué clase tiene más pasajeros registrados?

Utilizacion de contadores para mantener un registro de los usarios de cada clase

```python
c1, c2, c3 = 0, 0, 0

for item in df['Pclass']:
    if item == 1:
        c1 += 1
    elif item == 2:
        c2 += 1
    elif item == 3:
        c3 += 1

tabla_p2 = pd.DataFrame({
    "Clases": ["Primera", "Segunda", "Tercera"],
    "Total_Pasajeros": [c1, c2, c3]
})

display(tabla_p2)
```

# 3. ¿Cuántos pasajeros embarcaron en cada puerto y cuántos tienen el puerto sin registrar?

A traves de contadores para igualemnte tener referencias de los detalles de en que puertos embarcaron y el numero de tripulantes que subieron y se hace a traves del analisis de sus columnas correspondiente

```python
puerto_c, puerto_q, puerto_s, sin_puerto = 0, 0, 0, 0

for val in df['Embarked']:
    if val == 'C':
        puerto_c += 1
    elif val == 'Q':
        puerto_q += 1
    elif val == 'S':
        puerto_s += 1
    else:
        sin_puerto += 1

tabla_p3 = pd.DataFrame({
    "Puerto": ["Cherburgo", "Queenstown", "Southampton", "No Registrado"],
    "Pasajeros": [puerto_c, puerto_q, puerto_s, sin_puerto]
})

display(tabla_p3)
```

# 4. ¿Cuál es la tarifa promedio, mínima y máxima registrada para cada clase?

Se guarda los datos de cada lista para poder proceder al calculo del promedio

```python
precios_1 = []
precios_2 = []
precios_3 = []

for clase_val, tarifa_val in zip(df['Pclass'], df['Fare']):
    if clase_val == 1:
        precios_1.append(tarifa_val)
    elif clase_val == 2:
        precios_2.append(tarifa_val)
    elif clase_val == 3:
        precios_3.append(tarifa_val)

tabla_p4 = pd.DataFrame({
    "Medida": ["Promedio", "Máxima", "Mínima"],
    "Clase 1": [sum(precios_1)/len(precios_1), max(precios_1), min(precios_1)],
    "Clase 2": [sum(precios_2)/len(precios_2), max(precios_2), min(precios_2)],
    "Clase 3": [sum(precios_3)/len(precios_3), max(precios_3), min(precios_3)]
})

display(tabla_p4)
```

# 5. ¿Cuántos pasajeros de cada clase embarcaron en cada puerto?

Se analiza cada columna tanto de clase y puertos para poder llevar un conteo y examinar la cantidad de cada uno

```python
lista_c, lista_q, lista_s, lista_nan = [], [], [], []

for puerto_item, clase_item in zip(df['Embarked'], df['Pclass']):
    if puerto_item == 'C':
        lista_c.append(clase_item)
    elif puerto_item == 'Q':
        lista_q.append(clase_item)
    elif puerto_item == 'S':
        lista_s.append(clase_item)
    else:
        lista_nan.append(clase_item)

tabla_p5 = pd.DataFrame({
    "Clase_Pasajero": ["Clase 1", "Clase 2", "Clase 3"],
    "Cherburgo": [lista_c.count(1), lista_c.count(2), lista_c.count(3)],
    "Queenstown": [lista_q.count(1), lista_q.count(2), lista_q.count(3)],
    "Southampton": [lista_s.count(1), lista_s.count(2), lista_s.count(3)],
    "Sin Registro": [lista_nan.count(1), lista_nan.count(2), lista_nan.count(3)]
})

display(tabla_p5)
```

# 6. ¿Cuántos pasajeros sobrevivieron y cuántos fallecieron en cada clase?

Se busca a traves de un for si cumple que los pasajeros estan vivos o muertos para después contar cuantos vivos y muertos

```python
vivos = []
muertos = []

for estado_viva, clase_num in zip(df['Survived'], df['Pclass']):
    if estado_viva == 1:
        vivos.append(clase_num)
    else:
        muertos.append(clase_num)

tabla_p6 = pd.DataFrame({
    "Clase": ["Clase 1", "Clase 2", "Clase 3"],
    "Sobrevivientes": [vivos.count(1), vivos.count(2), vivos.count(3)],
    "Fallecidos": [muertos.count(1), muertos.count(2), muertos.count(3)]
})

display(tabla_p6)
```

# 7. ¿Cuántos pasajeros viajaban con cero, uno o dos o más familiares registrados a bordo?

Contadores de cada columna correspondiente del problema para poder trabajarla en conjunto

```python
acompanantes = df['SibSp'] + df['Parch']
solitarios = 0
un_familiar = 0
varios_familiares = 0

for num_fam in acompanantes:
    if num_fam == 0:
        solitarios += 1
    elif num_fam == 1:
        un_familiar += 1
    else:
        varios_familiares += 1

tabla_p7 = pd.DataFrame({
    "Grupo_Familiar": ["Sin acompañantes", "1 familiar", "2 o más familiares"],
    "Cantidad": [solitarios, un_familiar, varios_familiares]
})

display(tabla_p7)
```

# 8. ¿Cuántos pasajeros tenían menos de 18 años, cuántos tenían 18 o más y cuántos tenían la edad sin registrar? Desglose el resultado por clase.

A traves de un for se analiza las condiciones de cada columna para poder despues ponerla en el arreglo

```python
infantes = []
adultos = []
edades_nulas = []

for cl, ed in zip(df['Pclass'], df['Age']):
    if pd.isna(ed):
        edades_nulas.append(cl)
    elif ed < 18:
        infantes.append(cl)
    else:
        adultos.append(cl)

tabla_p8 = pd.DataFrame({
    "Clase": ["Clase 1", "Clase 2", "Clase 3"],
    "Menores de 18": [infantes.count(1), infantes.count(2), infantes.count(3)],
    "18 o más": [adultos.count(1), adultos.count(2), adultos.count(3)],
    "Sin Registrar": [edades_nulas.count(1), edades_nulas.count(2), edades_nulas.count(3)]
})

display(tabla_p8)
```

# 9. ¿Cuántos códigos de boleto distintos hay? ¿Cuántos aparecen asociados a más de un pasajero y cuál aparece en más registros?

Igualmente a traves de un for analizo las condiciones de los boletos para poder determinar la cantidad de cada uno

```python
frecuencias_ticket = df['Ticket'].value_counts()
total_distintos = len(frecuencias_ticket)

repetidos_multiples = 0
max_repeticiones = 0
ticket_estrella = ""

for tk in frecuencias_ticket.index:
    conteo_tk = frecuencias_ticket[tk]
    if conteo_tk > 1:
        repetidos_multiples += 1
    if conteo_tk > max_repeticiones:
        max_repeticiones = conteo_tk
        ticket_estrella = tk

tabla_p9 = pd.DataFrame({
    "Concepto": ["Boletos Únicos", "Boletos con más de un pasajero", "Boleto más repetido"],
    "Resultado": [total_distintos, repetidos_multiples, f"{ticket_estrella} ({max_repeticiones} veces)"]
})

display(tabla_p9)
```

# 10. ¿En qué clase hay más registros sin cabina? Para cada clase, muestren cuántos registros tienen cabina, cuántos carecen de ella y cuántos hay en total.

Utilizacion de bucles y condiciones para examinar las columnas de clase para ver si cuentan con cabina o no para poder despues trabajar en el arreglo

```python
con_habitacion = []
sin_habitacion = []

for cls, cab in zip(df['Pclass'], df['Cabin']):
    if pd.isna(cab):
        sin_habitacion.append(cls)
    else:
        con_habitacion.append(cls)

tabla_p10 = pd.DataFrame({
    "Clase": ["Clase 1", "Clase 2", "Clase 3"],
    "Con Cabina": [con_habitacion.count(1), con_habitacion.count(2), con_habitacion.count(3)],
    "Sin Cabina": [sin_habitacion.count(1), sin_habitacion.count(2), sin_habitacion.count(3)],
    "Total": [
        con_habitacion.count(1) + sin_habitacion.count(1),
        con_habitacion.count(2) + sin_habitacion.count(2),
        con_habitacion.count(3) + sin_habitacion.count(3)
    ]
})

display(tabla_p10)

sins_c1 = sin_habitacion.count(1)
sins_c2 = sin_habitacion.count(2)
sins_c3 = sin_habitacion.count(3)

if sins_c3 > sins_c1 and sins_c3 > sins_c2:
    print("Conclusión: La Clase 3 acumula la mayor cantidad de registros sin cabina.")
elif sins_c2 > sins_c1 and sins_c2 > sins_c3:
    print("Conclusión: La Clase 2 acumula la mayor cantidad de registros sin cabina.")
else:
    print("Conclusión: La Clase 1 acumula la mayor cantidad de registros sin cabina.")
```
