```python
import pandas as pd

productos = pd.read_csv('productos_.csv')

clientes = pd.read_csv('clientes.csv')

ventas = pd.read_csv('ventas.csv')

display(productos.head())

display(clientes.head())

display(ventas.head())
```

# 1. Líneas, tickets y unidades. ¿Cuántas líneas de venta, cuántos tickets distintos y cuántas unidades vendidas hay en el archivo? Explique qué representa cada conteo.

```python
lineas_venta = len(ventas)

# Cambio: se reemplazó set() por una lista [] y un if para evitar tickets repetidos
lista_tickets_unicos = []

for t in ventas['id_ticket']:
    if t not in lista_tickets_unicos:
        lista_tickets_unicos.append(t)

total_tickets = len(lista_tickets_unicos)

unidades_vendidas = 0

for cant in ventas['cantidad']:
    unidades_vendidas += cant

tabla_p1 = pd.DataFrame({
    "Métrica": ["Líneas de venta", "Tickets distintos", "Unidades vendidas"],
    "Descripción": [
        "Cada fila es un producto dentro de un ticket en ventas.csv",
        "Cantidad de compras o tickets diferentes sin repetir",
        "Total de unidades físicas vendidas sumando todas las filas"
    ],
    "Valor": [lineas_venta, total_tickets, unidades_vendidas]
})

display(tabla_p1)
```

# 2. Tickets por sucursal. ¿Cuántos tickets se registraron en cada sucursal? ¿Cuál registró más?

```python
tickets_por_sucursal = {}

# Cambio: en vez de set(), cada sucursal guarda una lista [] y solo agrega el ticket si no está ya en ella
for suc, tid in zip(ventas['sucursal'], ventas['id_ticket']):
    if suc not in tickets_por_sucursal:
        tickets_por_sucursal[suc] = []
    if tid not in tickets_por_sucursal[suc]:
        tickets_por_sucursal[suc].append(tid)

sucursales_l = []
cant_tickets_l = []

for suc, tids in tickets_por_sucursal.items():
    sucursales_l.append(suc)
    cant_tickets_l.append(len(tids))

tabla_p2 = pd.DataFrame({
    "Sucursal": sucursales_l,
    "Cantidad de Tickets": cant_tickets_l
})

display(tabla_p2)

max_t = -1
suc_max = ""

for suc, tids in tickets_por_sucursal.items():
    if len(tids) > max_t:
        max_t = len(tids)
        suc_max = suc

print(f"La sucursal que registró más tickets es: {suc_max} con {max_t} tickets.")
```

# 3. Métodos de pago. ¿Cuántos tickets se pagaron con cada método y cuántos tienen ese dato sin registrar?

```python
pago_por_ticket = {}

for metodo, tid in zip(ventas['metodo_pago'], ventas['id_ticket']):
    pago_por_ticket[tid] = metodo

conteo_metodos = {}

for tid, metodo in pago_por_ticket.items():
    # Cambio: se simplificó la etiqueta a "Sin registrar" para que sea más directa
    if pd.isna(metodo) or str(metodo).strip() == "":
        m_limpio = "Sin registrar"
    else:
        m_limpio = str(metodo)

    if m_limpio not in conteo_metodos:
        conteo_metodos[m_limpio] = 0
    conteo_metodos[m_limpio] += 1

metodos_l = []
conteo_l = []

for m, c in conteo_metodos.items():
    metodos_l.append(m)
    conteo_l.append(c)

tabla_p3 = pd.DataFrame({
    "Método de Pago": metodos_l,
    "Cantidad de Tickets": conteo_l
})

display(tabla_p3)
```

# 4. Precios de lista por categoría. ¿Cuál es el precio de lista mínimo, máximo y promedio de los productos de cada categoría? Use el catálogo de productos.

```python
precios_por_cat = {}

# Cambio: se agrega comentario indicando que se agrupan los precios de cada categoría
for cat, precio in zip(productos['categoria'], productos['precio_lista']):
    if cat not in precios_por_cat:
        precios_por_cat[cat] = []
    precios_por_cat[cat].append(precio)

cat_l = []
min_l = []
max_l = []
prom_l = []

for cat, precios in precios_por_cat.items():
    cat_l.append(cat)
    min_l.append(round(min(precios), 2))
    max_l.append(round(max(precios), 2))

    # Cambio: se suma con un bucle for en vez de usar sum() directo
    suma_p = 0
    for p in precios:
        suma_p += p
    prom_l.append(round(suma_p / len(precios), 2))

tabla_p4 = pd.DataFrame({
    "Categoría": cat_l,
    "Precio Mínimo ($)": min_l,
    "Precio Máximo ($)": max_l,
    "Precio Promedio ($)": prom_l
})

display(tabla_p4)
```

# 5. Unidades por sucursal y categoría. ¿Cuántas unidades vendió cada sucursal de cada categoría? Presente una tabla cruzada con totales por fila y por columna. Compruebe que el total general coincide con la suma de cantidad en ventas.csv.

```python
# Cambio: se crea un diccionario para relacionar cada id_producto con su categoría
prod_a_cat = {}

for pid, cat in zip(productos['id_producto'], productos['categoria']):
    prod_a_cat[pid] = cat

cats_ventas = []

for pid in ventas['id_producto']:
    cats_ventas.append(prod_a_cat.get(pid, "Desconocido"))

ventas['categoria'] = cats_ventas

tabla_unidades = pd.crosstab(
    ventas['sucursal'],
    ventas['categoria'],
    values=ventas['cantidad'],
    aggfunc='sum',
    margins=True,
    margins_name='Total'
)

display(tabla_unidades)

total_tabla = tabla_unidades.loc['Total', 'Total']

# Cambio: se suma cantidad fila por fila con un for para comprobar el total general
suma_manual_cant = 0

for cant in ventas['cantidad']:
    suma_manual_cant += cant

print(f"Total en la tabla cruzada: {total_tabla}")

print(f"Suma manual en ventas.csv: {suma_manual_cant}")

print(f"¿Coinciden los totales?: {total_tabla == suma_manual_cant}")
```

# 6. Importe de ventas por sucursal. ¿Cuál fue el importe total de ventas de cada sucursal? Calcule primero el precio total de cada línea como cantidad × precio_unitario.

```python
importe_por_sucursal = {}

# Cambio: se comenta el cálculo de cada línea (cantidad * precio_unitario) acumulado por sucursal
for suc, cant, precio in zip(ventas['sucursal'], ventas['cantidad'], ventas['precio_unitario']):
    linea_importe = cant * precio
    if suc not in importe_por_sucursal:
        importe_por_sucursal[suc] = 0.0
    importe_por_sucursal[suc] += linea_importe

suc_l = []
imp_l = []

for suc, imp in importe_por_sucursal.items():
    suc_l.append(suc)
    imp_l.append(round(imp, 2))

tabla_p6 = pd.DataFrame({
    "Sucursal": suc_l,
    "Importe Total ($)": imp_l
})

display(tabla_p6)
```

# 7. Tamaño de las compras. ¿Cuántos tickets corresponden a compras de hasta 7 unidades, de 8 a 14 unidades y de 15 o más unidades? Sume primero las cantidades de todas las líneas de cada ticket.

```python
ticket_cantidades = {}

# Cambio: primero se suman las cantidades de todas las líneas de cada ticket
for tid, cant in zip(ventas['id_ticket'], ventas['cantidad']):
    if tid not in ticket_cantidades:
        ticket_cantidades[tid] = 0
    ticket_cantidades[tid] += cant

hasta_7 = 0
de_8_a_14 = 0
mas_15 = 0

# Cambio: se clasifican los tickets con if / elif / else según su total de unidades
for tid, total_c in ticket_cantidades.items():
    if total_c <= 7:
        hasta_7 += 1
    elif total_c <= 14:
        de_8_a_14 += 1
    else:
        mas_15 += 1

tabla_p7 = pd.DataFrame({
    "Tamaño de Compra": ["Hasta 7 unidades", "De 8 a 14 unidades", "15 o más unidades"],
    "Cantidad de Tickets": [hasta_7, de_8_a_14, mas_15]
})

display(tabla_p7)
```

# 8. Ventas por mes. ¿Cuántos tickets y qué importe total se registraron en cada mes? ¿En qué mes se alcanzó el mayor importe?

```python
meses_importes = {}
meses_tickets = {}

for fecha, tid, cant, precio in zip(ventas['fecha'], ventas['id_ticket'], ventas['cantidad'], ventas['precio_unitario']):
    mes = str(fecha)[:7]
    importe = cant * precio

    # Cambio: se inicializa meses_tickets[mes] como lista [] en lugar de set()
    if mes not in meses_importes:
        meses_importes[mes] = 0.0
        meses_tickets[mes] = []

    meses_importes[mes] += importe

    # Cambio: se verifica con if para no contar el mismo ticket dos veces en el mismo mes
    if tid not in meses_tickets[mes]:
        meses_tickets[mes].append(tid)

meses_l = []
tickets_l = []
importes_l = []

for m in sorted(meses_importes.keys()):
    meses_l.append(m)
    tickets_l.append(len(meses_tickets[m]))
    importes_l.append(round(meses_importes[m], 2))

tabla_p8 = pd.DataFrame({
    "Mes": meses_l,
    "Tickets Distintos": tickets_l,
    "Importe Total ($)": importes_l
})

display(tabla_p8)

max_imp = -1
mes_mayor = ""

for m, imp in meses_importes.items():
    if imp > max_imp:
        max_imp = imp
        mes_mayor = m

print(f"El mes que alcanzó el mayor importe de ventas fue: {mes_mayor} con ${round(max_imp, 2)}")
```

# 9. Producto más vendido por sucursal. ¿Qué producto acumuló más unidades vendidas en cada sucursal? Indique su código, nombre y cantidad vendida. Si hay un empate, incluya todos los productos empatados.

```python
id_a_nombre = {}
for pid, nom in zip(productos['id_producto'], productos['producto']):
    id_a_nombre[pid] = nom

# Cambio: sumar unidades vendidas por cada par (sucursal, id_producto)
acum_suc_prod = {}
for suc, pid, cant in zip(ventas['sucursal'], ventas['id_producto'], ventas['cantidad']):
    clave = (suc, pid)
    if clave not in acum_suc_prod:
        acum_suc_prod[clave] = 0
    acum_suc_prod[clave] += cant

# Cambio: buscar el mayor número de unidades vendidas en cada sucursal
max_por_suc = {}
for (suc, pid), cant in acum_suc_prod.items():
    if suc not in max_por_suc:
        max_por_suc[suc] = cant
    else:
        if cant > max_por_suc[suc]:
            max_por_suc[suc] = cant

# Cambio: incluir todos los productos que lleguen al máximo (sirve para empates)
resultados_p9 = []
for (suc, pid), cant in acum_suc_prod.items():
    if cant == max_por_suc[suc]:
        resultados_p9.append({
            "Sucursal": suc,
            "Código Producto": pid,
            "Nombre Producto": id_a_nombre.get(pid, "Desconocido"),
            "Cantidad Vendida": cant
        })

tabla_p9 = pd.DataFrame(resultados_p9)
display(tabla_p9)
```

# 10. Compras por tipo de cliente. Para cada tipo de cliente, ¿cuántos clientes distintos compraron, cuántos tickets generaron y cuál fue el importe total de sus compras?

```python
cliente_a_tipo = {}
for cid, tipo in zip(clientes['id'], clientes['tipo_cliente']):
    cliente_a_tipo[cid] = tipo

clientes_por_tipo = {}
tickets_por_tipo = {}
importe_por_tipo = {}

for cid, tid, cant, precio in zip(ventas['id_cliente'], ventas['id_ticket'], ventas['cantidad'], ventas['precio_unitario']):
    tipo = cliente_a_tipo.get(cid, "Desconocido")
    importe = cant * precio

    # Cambio: se usan listas [] en vez de set() para guardar clientes y tickets únicos por tipo
    if tipo not in importe_por_tipo:
        clientes_por_tipo[tipo] = []
        tickets_por_tipo[tipo] = []
        importe_por_tipo[tipo] = 0.0

    # Cambio: se agrega cid y tid solo si aún no estaban en la lista de ese tipo de cliente
    if cid not in clientes_por_tipo[tipo]:
        clientes_por_tipo[tipo].append(cid)
    if tid not in tickets_por_tipo[tipo]:
        tickets_por_tipo[tipo].append(tid)

    importe_por_tipo[tipo] += importe

tipos_l = []
dist_l = []
gen_l = []
imp_l = []

for tipo in sorted(importe_por_tipo.keys()):
    tipos_l.append(tipo)
    dist_l.append(len(clientes_por_tipo[tipo]))
    gen_l.append(len(tickets_por_tipo[tipo]))
    imp_l.append(round(importe_por_tipo[tipo], 2))

tabla_p10 = pd.DataFrame({
    "Tipo de Cliente": tipos_l,
    "Clientes Distintos": dist_l,
    "Tickets Generados": gen_l,
    "Importe Total ($)": imp_l
})

display(tabla_p10)
```
